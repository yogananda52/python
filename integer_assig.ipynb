Question 5: Calculate the sum of digits in number 12345

```python
num=12345
temp=num
sum_of=0
while num>0:
    sum_of+=int(num%10)
    num/=10
print(f"sum of digits in num {temp} is {sum_of} ")
```

Question 6: Check if 97 is a prime number

```python
num=97
prime=True
for i in range(2,int(num/2)):
    if num%i==0:
        prime=False
if prime:
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
    
```

 Question 7: Find the factorial of 8

Question 9: Find the greatest common divisor (GCD) of 48 and 36

```python
value_1=48
value_2=30
Gcd=1

for i in range(1,int(min(value_1,value_2)/2)):
    if value_1%i==0 and value_2%i==0:
        Gcd=i
print(f"GCD of {value_1} and {value_2} is {Gcd}")
```

Question 10: Calculate the sum of first 20 odd numbers

```python
sum_of_odd=0
count=1
num=1
while count<=20:
    if num%2!=0:
        count+=1
        sum_of_odd+=num
    num+=1
print(f"sum of first 20 odd nums is {sum_of_odd}")
        

```
