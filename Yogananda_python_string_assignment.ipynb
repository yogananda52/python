## Question 41

Check if string contains balanced parentheses: "((()))"

```python
s = "((()))"
depth = 0
balanced = True
for ch in s:
    if ch == "(":
        depth += 1
    elif ch == ")":
        depth -= 1
        if depth < 0:
            balanced = False
            break
print(balanced and depth == 0)
```

## Question 43

Find the longest common substring between "programming" and "grammar"

```python
a, b = "programming", "grammar"
best = ""
for i in range(len(a)):
    for j in range(i + 1, len(a) + 1):
        candidate = a[i:j]
        if candidate in b and len(candidate) > len(best):
            best = candidate
print(best)
```

## Question 46

Convert "hello world" to Pig Latin

```python
def pig_latin(word):
    vowels = "aeiou"
    for i, ch in enumerate(word):
        if ch in vowels:
            return word[i:] + word[:i] + "ay"
    return word + "ay"
s = "hello world"
print(" ".join(pig_latin(word) for word in s.split()))
```

## Question 50

Check if string is a valid credit card number: "4532015112830366"

```python
def luhn_check(number):
    digits = [int(ch) for ch in number if ch.isdigit()]
    if len(digits) != len(number) or not 13 <= len(digits) <= 19:
        return False
    total = 0
    for i, digit in enumerate(reversed(digits)):
        if i % 2 == 1:
            digit *= 2
            if digit > 9:
                digit -= 9
        total += digit
    return total % 10 == 0

s = "4532015112830366"
print(luhn_check(s))  # Checks number format/checksum, not whether the card is active
```
