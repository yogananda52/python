## Question 12

```python
first = "listen"
second = "silent"

if len(first) != len(second):
    print(False)
else:
    checked = ""
    is_anagram = True

    for char in first:
        if char not in checked:
            count_first = 0
            count_second = 0

            for item in first:
                if item == char:
                    count_first += 1
            for item in second:
                if item == char:
                    count_second += 1

            if count_first != count_second:
                is_anagram = False
                break
            checked += char

    print(is_anagram)
```

## Question 23

Convert "hello world" to "hElLo WoRlD" (alternating case)

```python
text = "hello world"
result = ""
position = 0

for char in text:
    if char == " ":
        result += char
    else:
        if position % 2 == 0:
            if "A" <= char <= "Z":
                result += chr(ord(char) + 32)
            else:
                result += char
        else:
            if "a" <= char <= "z":
                result += chr(ord(char) - 32)
            else:
                result += char
        position += 1

print(result)
```

## Question 42

Convert "hello world" to Morse code

```python
text = "hello world"
letters = "abcdefghijklmnopqrstuvwxyz"
codes = [".-", "-...", "-.-.", "-..", ".", "..-.", "--.", "....", "..", ".---",
         "-.-", ".-..", "--", "-.", "---", ".--.", "--.-", ".-.", "...", "-",
         "..-", "...-", ".--", "-..-", "-.--", "--.."]
result = ""

for char in text:
    if char == " ":
        result += "/ "
    else:
        position = -1
        for i in range(len(letters)):
            if char == letters[i]:
                position = i
                break
        if position != -1:
            result += codes[position] + " "

print(result)
```

## Question 43

Find the longest common substring between "programming" and "grammar"

## Question 46

Convert "hello world" to Pig Latin

```python
text = "hello world"
words = []
word = ""

for char in text + " ":
    if char != " ":
        word += char
    else:
        if word != "":
            vowel_position = -1
            for i in range(len(word)):
                if word[i] in "aeiou":
                    vowel_position = i
                    break

            if vowel_position == -1:
                pig_word = word + "ay"
            else:
                pig_word = word[vowel_position:] + word[:vowel_position] + "ay"

            words.append(pig_word)
            word = ""

print(words[0] + " " + words[1])
```

## Question 47

Check if string is a valid IPv4 address: "192.168.1.1"

```python
text = "192.168.1.1"
parts = []
part = ""

for char in text:
    if char == ".":
        parts.append(part)
        part = ""
    else:
        part += char
parts.append(part)

valid = len(parts) == 4

for part in parts:
    if part == "" or len(part) > 3:
        valid = False
        break

    number = 0
    for char in part:
        if char < "0" or char > "9":
            valid = False
            break
        number = number * 10 + (ord(char) - ord("0"))

    if number > 255:
        valid = False

print(valid)  # Beginner-level IPv4 check
```

## Question 49

Convert "hello world" to ROT13 encoding

```python
text = "hello world"
result = ""

for char in text:
    if "a" <= char <= "z":
        number = ord(char) - ord("a")
        number = (number + 13) % 26
        result += chr(ord("a") + number)
    elif "A" <= char <= "Z":
        number = ord(char) - ord("A")
        number = (number + 13) % 26
        result += chr(ord("A") + number)
    else:
        result += char

print(result)
```

## Question 50

Check if string is a valid credit card number: "4532015112830366"

```python
number = "4532015112830366"
total = 0
double_digit = False
valid_digits = True

for i in range(len(number) - 1, -1, -1):
    char = number[i]

    if char < "0" or char > "9":
        valid_digits = False
        break

    digit = ord(char) - ord("0")

    if double_digit:
        digit = digit * 2
        if digit > 9:
            digit -= 9

    total += digit
    double_digit = not double_digit

print(valid_digits and total % 10 == 0)  # Checks checksum only, not whether a card is active
```
