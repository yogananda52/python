### Problem 1.5: Calculate the sum of digits in number 12345

```python
num = 12345
temp = num
sum_of = 0
while num > 0:
    sum_of += num % 10
    num //= 10
print(f"sum of digits in num {temp} is {sum_of}")
```

### Problem 1.6: Check if 97 is a prime number

```python
num = 97
prime = True
if num < 2:
    prime = False
else:
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            prime = False
            break
if prime:
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```

### Problem 1.9: Find the greatest common divisor (GCD) of 48 and 36

```python
value_1 = 48
value_2 = 36
gcd_value = 1

for i in range(1, min(value_1, value_2) + 1):
    if value_1 % i == 0 and value_2 % i == 0:
        gcd_value = i
print(f"GCD of {value_1} and {value_2} is {gcd_value}")
```

### Problem 1.10: Calculate the sum of first 20 odd numbers

```python
sum_of_odd=0
count=1
num=1
while count<=20:
    if num%2!=0:
        count+=1
        sum_of_odd+=num
    num+=1
print(f"sum of first 20 odd nums is {sum_of_odd}")


```

### Problem 2.2: Convert 98.6 Fahrenheit to Celsius (F = C * 9/5 + 32)

```python
fahrenheit=98.6
celsius=(fahrenheit-32)*5/9
print(f"{fahrenheit} in celsius is {celsius}")
```

### Problem 2.3: Calculate the compound interest on $1000 at 5.5% for 3 years

```python
principal = 1000
rate = 5.5
years = 3

amount = principal * (1 + rate / 100) ** years
compound_interest = amount - principal

print(f"Compound interest: ${compound_interest:.2f}")
```

### Problem 2.4: Find the hypotenuse of a right triangle with sides 3.5 and 4.2

```python
base = 3.5
height = 4.2

hypotenuse = (base ** 2 + height ** 2) ** 0.5
print(f"The hypotenuse is {hypotenuse:.2f}")
```

### Problem 2.5: Calculate the volume of a sphere with radius 7.8

```python
import math

radius = 7.8
volume = (4 / 3) * math.pi * radius ** 3

print(f"The volume of the sphere is {volume:.2f}")
```
