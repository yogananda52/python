## Question 4

Convert "hello world" to title case

```python
text = "hello world"
result = ""
make_upper = True

for char in text:
    if char == " ":
        result += char
        make_upper = True
    elif make_upper:
        result += chr(ord(char) - 32) if "a" <= char <= "z" else char
        make_upper = False
    else:
        result += char

print(result)
```

## Question 39

Find the most common word in "the quick brown fox jumps over the lazy dog"

```python
text = "the quick brown fox jumps over the lazy dog"
words = []
word = ""

for char in text + " ":
    if char != " ":
        word += char
    else:
        if word != "":
            words.append(word)
            word = ""

most_common = words[0]
highest_count = 0

for candidate in words:
    count = 0
    for other in words:
        if candidate == other:
            count += 1
    if count > highest_count:
        highest_count = count
        most_common = candidate

print(f"The most common word in the text is '{most_common}' which occurs {highest_count} times.") 
```

## Question 41

Check if string contains balanced parentheses: "((()))"

```python
text = "((()))"
balance = 0
valid = True

for char in text:
    if char == "(":
        balance += 1
    elif char == ")":
        balance -= 1

    if balance < 0:
        valid = False
        break

print(f"Output: {valid and balance == 0}")
```

## Question 42

Convert "hello world" to Morse code

```python
text = "hello world"
letters = "abcdefghijklmnopqrstuvwxyz"
codes = [".-", "-...", "-.-.", "-..", ".", "..-.", "--.", "....", "..", ".---",
         "-.-", ".-..", "--", "-.", "---", ".--.", "--.-", ".-.", "...", "-",
         "..-", "...-", ".--", "-..-", "-.--", "--.."]
result = ""

for char in text:
    if char == " ":
        result += "/ "
    else:
        position = -1
        for i in range(len(letters)):
            if char == letters[i]:
                position = i
                break
        if position != -1:
            result += codes[position] + " "

print(f"Output: {result}")
```

## Question 43

Find the longest common substring between "programming" and "grammar"

```python
first = "programming"
second = "grammar"
longest = ""

for start in range(len(first)):
    current = ""
    for end in range(start, len(first)):
        current += first[end]

        found = False
        for i in range(len(second) - len(current) + 1):
            match = True
            for j in range(len(current)):
                if second[i + j] != current[j]:
                    match = False
                    break
            if match:
                found = True
                break

        if found and len(current) > len(longest):
            longest = current

print(f"Output: {longest}")
```

## Question 46

Convert "hello world" to Pig Latin

```python
text = "hello world"
words = []
word = ""

for char in text + " ":
    if char != " ":
        word += char
    else:
        if word != "":
            vowel_position = -1
            for i in range(len(word)):
                if word[i] in "aeiou":
                    vowel_position = i
                    break

            if vowel_position == -1:
                pig_word = word + "ay"
            else:
                pig_word = word[vowel_position:] + word[:vowel_position] + "ay"

            words.append(pig_word)
            word = ""

print(f"Output: {words[0] + " " + words[1]}")
```

## Question 47

Check if string is a valid IPv4 address: "192.168.1.1"

```python
text = "192.168.1.1"
parts = []
part = ""

for char in text:
    if char == ".":
        parts.append(part)
        part = ""
    else:
        part += char
parts.append(part)

valid = len(parts) == 4

for part in parts:
    if part == "" or len(part) > 3:
        valid = False
        break

    number = 0
    for char in part:
        if char < "0" or char > "9":
            valid = False
            break
        number = number * 10 + (ord(char) - ord("0"))

    if number > 255:
        valid = False

print(valid)  # Beginner-level IPv4 check
```

## Question 49

Convert "hello world" to ROT13 encoding

```python
text = "hello world"
result = ""

for char in text:
    if "a" <= char <= "z":
        number = ord(char) - ord("a")
        number = (number + 13) % 26
        result += chr(ord("a") + number)
    elif "A" <= char <= "Z":
        number = ord(char) - ord("A")
        number = (number + 13) % 26
        result += chr(ord("A") + number)
    else:
        result += char

print(f"Output: {result}")
```

## Question 50

Check if string is a valid credit card number: "4532015112830366"

```python
number = "4532015112830366"
total = 0
double_digit = False
valid_digits = True

for i in range(len(number) - 1, -1, -1):
    char = number[i]

    if char < "0" or char > "9":
        valid_digits = False
        break

    digit = ord(char) - ord("0")

    if double_digit:
        digit = digit * 2
        if digit > 9:
            digit -= 9

    total += digit
    double_digit = not double_digit

print(valid_digits and total % 10 == 0)  # Checks checksum only, not whether a card is active
```
