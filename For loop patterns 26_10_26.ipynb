# 1. Right-angled triangle — stars

```python
n=5
for i in range(1,n):
    print('*'*i)
```

# 2. Inverted right-angled triangle — stars

```python
n=5
for i in range(n,0,-1):
    print('*'*i)
```

# 3. Solid square of stars

```python
n=5
for i in range(1,n+1):
    print('* '*n)
```

# 4. Hollow square

```python
n=5
for i in range(n+1):
    for j in range(n+1):
        if i ==0 or j==0 or i==n or j==n:
            print("*",end="")
        else:
            print(" ",end="")
    print()
```

# 5. Number triangle (1, 12, 123, …)

```python
n=5
for i in range (n+1):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

# 6. Floyd’s triangle (continuous numbers)

```python
n=5
count=1
for i in range (n+1):
    for j in range(1,i+1):
        print(count,end=" ")
        count+=1
    print()
```

# 7. Centered star pyramid

```python
n=6
for i in range(n+1):
    print(" " * (n - i) + "*" * (2 * i - 1))
   

```

 # 8. Inverted centered star pyramid

```python
n=7
for i in range(n,0,-1):
    print(" " *(n-i) +"*"*(i*2-1))
```

# 9. Diamond (centered, stars)

```python
n=5
for i in range(n+1):
    print(" "*(n-i)+"*"*(i*2-1))
    if i==n:
        for i in range(n-1,0,-1):
            print(" "*(n-i)+"*"*(i*2-1))
```

# 10. Same digit per row

```python
n=5
for i in range(1,n+1):
    for j in range(1,i+1):
        print(i,end="")
    print()
```

# 11. Binary triangle (0 and 1)

```python
n=5
for i in range(1,n+1):
    for j in range(1,i+1):
        if j%2==0:
            print("1",end="")
        else:
            print("0",end="")
    print()
```

# 12. Checkerboard


```python
n=4
for i in range(n+1):
    for j in range(n+1):
        if i+j%2==0:
            print("# ",end="")
        else:
            print(". ",end="")
    print()
```

# 14. Alphabet right angletriangle

```python
n=5
for i in range(n):
    for j in range(i+1):
        print(chr(ord("A")+j),end="")
    print()
```
