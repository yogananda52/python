## Problem 4.3: Pattern Printing - Pyramid (Medium)

Print a centered pyramid pattern for 5 rows:

```text
    *
   ***
  *****
 *******
*********
```

For each row, print the required leading spaces followed by an odd number of stars.

````python
n = 5
for row in range(1, n + 1):
    spaces = n - row
    stars = 2 * row - 1
    print(" " * spaces + "*" * stars)
````

## Problem 4.4: Pattern Printing - Number Pattern (Medium)

Print the following pattern for 5 rows:

```text
1
12
123
1234
12345
```

Each row contains the numbers from 1 up to its row number.

````python
n = 5
for row in range(1, n + 1):
    for number in range(1, row + 1):
        print(number, end="")
    print()
````

## Problem 4.5: Multiplication Table Grid (Medium)

Print a multiplication table grid from 1 to 5, with aligned columns.

````python
n = 5
for row in range(1, n + 1):
    for column in range(1, n + 1):
        print(f"{row * column:4}", end="")
    print()
````

## Problem 4.6: Diamond Pattern (Hard)

Print a diamond with 9 rows (the number of rows should be odd):

```text
    *
   ***
  *****
 *******
*********
 *******
  *****
   ***
    *
```

Print the upper half, including the middle row, then the lower half.

````python
n = 9
if n % 2 == 0:
    raise ValueError("The number of rows must be odd.")
middle = n // 2 + 1
for row in range(1, middle + 1):
    print(" " * (middle - row) + "*" * (2 * row - 1))
for row in range(middle - 1, 0, -1):
    print(" " * (middle - row) + "*" * (2 * row - 1))
````

## Section 5: Break and Continue Statements

### Problem 5.1: Find First Multiple (Easy)

Find and print the first number from 1 to 100 that is divisible by both 7 and 13. Use `break` to exit the loop once it is found.

````python
for number in range(1, 101):
    if number % 7 == 0 and number % 13 == 0:
        print(f"First multiple of both 7 and 13: {number}")
        break
````

### Problem 5.2: Skip Multiples (Easy)

Print all numbers from 1 to 50, excluding multiples of 3. Use `continue` to skip them.

````python
for number in range(1, 51):
    if number % 3 == 0:
        continue
    print(number)
````

### Problem 5.3: Sum Until Negative (Medium)

Repeatedly ask the user to enter an integer. Add positive numbers to a running total, skip zero with `continue`, and stop when a negative number is entered using `break`. Print the final sum.

````python
total = 0
while True:
    number = int(input("Enter an integer: "))
    if number < 0:
        break
    if number == 0:
        continue
    total += number
print(f"Sum of positive numbers: {total}")
````
